```python
import pandas as pd

test = pd.read_csv("telemetry_test.csv")
SENS = [c for c in test.columns if c not in ["timestamp_id", "session_id", "mode", "dt"]]
```

Считаем, насколько показание датчика отклонилось от его обычного уровня
в этом сеансе связи, и усредняем по датчикам.

```python
score = 0
for s in SENS:
    x = test[s].ffill().fillna(0)
    usual = x.groupby(test.session_id).transform("median")   # обычный уровень датчика в сеансе
    score += (x - usual).abs() / (x.std() + 1e-9)             # общий масштаб для всех датчиков
score /= len(SENS)

sub = pd.DataFrame({"timestamp_id": test.timestamp_id, "anomaly_prob": score})
sub.to_csv("submission.csv", index=False)
sub.head()
```
